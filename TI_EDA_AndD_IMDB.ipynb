```python
import kagglehub
import os
import pandas as pd
import numpy as np

path = kagglehub.dataset_download("asaniczka/tmdb-movies-dataset-2023-930k-movies")
print("Path to dataset files:", path)

for f in os.listdir(path):
    print(f)

csv_path = os.path.join(path, "TMDB_movie_dataset_v11.csv")
movies_df = pd.read_csv(csv_path, low_memory=False)

```

```python
print(f"Dimensiones del dataset original: {movies_df.shape}")
movies_df.head()
```

```python
movies_df.info()
```

```python
nulos = movies_df.isnull().sum()
porcentaje = (nulos / len(movies_df) * 100).round(2)

resumen_nulos = pd.DataFrame({
    'Nulos': nulos,
    'Porcentaje (%)': porcentaje
}).sort_values(by='Porcentaje (%)', ascending=False)

resumen_nulos[resumen_nulos['Nulos'] > 0]
```

```python
# Fechas
movies_df['release_date'] = pd.to_datetime(movies_df['release_date'], errors='coerce')

# Categóricas de baja cardinalidad
for col in ['status', 'original_language']:
    movies_df[col] = movies_df[col].astype('category')
```
